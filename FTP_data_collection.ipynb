```python
import pandas as pd

#import nifty500 listed stocks
url1 = 'https://en.wikipedia.org/wiki/NIFTY_500'
nifty500 = pd.read_html(url1)[4]

# If columns are not named, we assign them manually
expected_columns = ['Sl.No', 'Company Name', 'Industry', 'Symbol', 'Series', 'ISIN Code']
if list(nifty500.columns) != expected_columns:
    nifty500.columns = expected_columns

# Extract the 'Symbol' column and add '.NS' for Nifty listed stocks
nifty500['Symbol'] = nifty500['Symbol'].str.strip() + '.NS'

symbols_list1 = nifty500['Symbol'].tolist()[1:]

print(symbols_list1)
print(len(symbols_list1))
```

```python
nifty500 = nifty500.drop(index=0).reset_index(drop=True)
```

```python
nifty500.drop(columns=['Industry', 'Series', 'ISIN Code'], inplace=True)
```

```python
nifty500.head()
```

```python
#import S&P 500 list of stocks
url2 = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
sp500 = pd.read_html(url2)[0]

# If columns are not named, we assign them manually
expected_columns = ['Symbol', 'Security', 'GICS Sector', 'GICS Sub-Industry', 'Headquarters Location', 'Date added', 'CIK', 'Founded']
if list(sp500.columns) != expected_columns:
    sp500.columns = expected_columns

# Now extract the 'Symbol' column
sp500['Symbol'] = sp500['Symbol'].str.strip()

symbols_list2 = sp500['Symbol'].tolist()
symbols_list2 = [ticker.replace('.', '-') for ticker in symbols_list2]

print(symbols_list2)
print(len(symbols_list2))
```

```python
sp500.insert(0, 'Sl.No', range(1, len(sp500) + 1))
```

```python
sp500.rename(columns={'Security': 'Company Name'}, inplace=True)
sp500.drop(columns=['GICS Sector', 'GICS Sub-Industry', 'Headquarters Location', 'Date added', 'CIK', 'Founded'], inplace=True)
```

```python
sp500.head()
```

```python
# unified table (Nifty 500 and S&P 500)
data_uni = pd.concat([nifty500, sp500], ignore_index=True)
data_uni.head(1030)
```

```python
symbols_list = data_uni['Symbol'].tolist()
print(symbols_list)
print(len(symbols_list))
```

```python
import yfinance as yf
import numpy as np
```

```python
help(yf.download)
```

```python
# We will use 5-year window to collect closed prices of listed stocks on both exchanges
end_date = '2025-05-31'
start_date = '2020-01-01'

# Download data from Yahoo Finance
data_uni = yf.download(
    tickers=symbols_list,
    start=start_date,
    end=end_date,
    interval='1d',
    group_by='ticker',
    auto_adjust=False,
    threads=True
)

# Extract Close prices (handle missing tickers safely)
close_prices = {}
for sym in symbols_list:
    try:
        close = data_uni.get(sym, pd.DataFrame()).get('Close')
        if close is not None:
            close_prices[sym] = close
    except Exception as e:
        print(f"Skipping {sym}: {e}")

# Create final DataFrame
close_df = pd.DataFrame(close_prices).ffill().dropna(axis=1)

# Preview
print(close_df.head())
print(f"Number of stocks with complete data: {close_df.shape[1]}")
```

```python
close_df.head()
```

```python
close_df.to_csv('closed_prices.csv', index=True, date_format='%Y-%m-%d')
print("Saved closed_prices.csv")
```

```python
print(f"Data shape: {close_df.shape} (Days × Stocks)")
```

```python
# Calculate daily returns (percentage change)

returns_df = close_df.pct_change().dropna()
```

```python
returns_df.to_csv('daily_returns.csv')
print("Saved daily_returns.csv")
```
